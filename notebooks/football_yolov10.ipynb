```python
import os
import torch
import shutil
import yaml



!pip install -q "numpy<2.0"
!pip install -q git+https://github.com/THU-MIG/yolov10.git
!pip uninstall -y wandb

from ultralytics import YOLOv10


if not os.path.exists('/content/yolov10n.pt'):
    !wget -q https://github.com/THU-MIG/yolov10/releases/download/v1.1/yolov10n.pt

os.environ['KAGGLE_USERNAME'] = "btammm"
os.environ['KAGGLE_KEY'] = "KGAT_6e6c6a015976401306d8a7c9af925e6e"

if not os.path.exists('/content/football_data'):
    !kaggle datasets download -d iasadpanwhar/football-player-detection-yolov8 -p /content
    !unzip -q /content/football-player-detection-yolov8.zip -d /content/football_data


src = '/content/football_data/football_players_detection/football_players_detection'
dest = '/content/mini_final'
if os.path.exists(dest): shutil.rmtree(dest)
for p in ['train/images', 'train/labels', 'valid/images', 'valid/labels']:
    os.makedirs(os.path.join(dest, p), exist_ok=True)

def copy_data(folder, n):
    imgs = sorted(os.listdir(os.path.join(src, folder, 'images')))[:n]
    for f in imgs:
        shutil.copy(os.path.join(src, folder, 'images', f), os.path.join(dest, folder, 'images', f))
        lbl = f.rsplit('.', 1)[0] + '.txt'
        if os.path.exists(os.path.join(src, folder, 'labels', lbl)):
            shutil.copy(os.path.join(src, folder, 'labels', lbl), os.path.join(dest, folder, 'labels', lbl))

copy_data('train', 300)
copy_data('valid', 100)

with open('/content/mini.yaml', 'w') as f:
    yaml.dump({'train': f'{dest}/train/images', 'val': f'{dest}/valid/images', 'nc': 2, 'names': ['football', 'player']}, f)


import torch.serialization
if not hasattr(torch, '_original_load'): torch._original_load = torch.load
torch.load = lambda *args, **kwargs: torch._original_load(*args, **{**kwargs, 'weights_only': False})


try:
    model = YOLOv10('/content/yolov10n.pt')


    model.callbacks = {}

    model.train(
        data='/content/mini.yaml',
        epochs=10,
        batch=32,
        imgsz=320,
        device=0,
        cache=True,
        plots=False
    )

finally:
    torch.load = torch._original_load
```
